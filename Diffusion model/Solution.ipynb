```python
from transformers import AutoTokenizer, AutoModelForMaskedLM
```

```python
tokenizer = AutoTokenizer.from_pretrained("tommyp111/modernbert-diffusion")
model = AutoModelForMaskedLM.from_pretrained("tommyp111/modernbert-diffusion")
```

```python
input_text = "The future of AI is [MASK]"
```

```python
inputs = tokenizer(input_text, return_tensors="pt")
inputs
```

```python
outputs = model(**inputs)
outputs
```

```python
logits = outputs.logits
logits
```

```python
mask_token_index = (inputs.input_ids == tokenizer.mask_token_id)[0].nonzero(as_tuple=True)[0]
mask_token_index
```

```python
predicted_token_id = logits[0, mask_token_index].argmax(axis=-1)
```

```python
predicted_token = tokenizer.decode(predicted_token_id)
predicted_token
```

```python
from transformers import pipeline
unmasker = pipeline('fill-mask', model='tommyp111/modernbert-diffusion')
unmasker("The future of AI is [MASK]")
```
